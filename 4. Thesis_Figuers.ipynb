```python
# =============================================================
# CELL 1 — SETUP
# PURPOSE:
#   Define input/output paths, helper functions, and safe file naming.
#   This cell is required once before running the plotting cells.
# =============================================================

import os
import arcpy
import matplotlib.pyplot as plt

# --- Base directories (edit only if needed) ---
base_dir = r"C:\Users\SHIDEBNA\Downloads\Thesis_Analysis\WindDirection_Output"
ten_year_dir   = os.path.join(base_dir, "ten_year_mean")
decadal_dir    = os.path.join(base_dir, "decadal_change")
multi_dir      = os.path.join(base_dir, "multi_interval_change")

# --- Output root for the 5 columns ---
fig_root = os.path.join(base_dir, "FIGURES_5_COLUMNS")
os.makedirs(fig_root, exist_ok=True)

# Column folders
col1 = os.path.join(fig_root, "1_Setups_Documentation")
col2 = os.path.join(fig_root, "2_TenYear_Line_PerDecade")
col3 = os.path.join(fig_root, "3_TenYear_Combined_AllDecades")
col4 = os.path.join(fig_root, "4_Change_Continuous_Distributions")
col5 = os.path.join(fig_root, "5_MedianAbsTrend_ByBaselineDecade")

for p in [col1, col2, col3, col4, col5]:
    os.makedirs(p, exist_ok=True)

# Seasons and decades (match your file names)
seasons_list = ["DJF", "MAM", "JJA", "SON"]
decades = ["1985-1994", "1995-2004", "2005-2014", "2015-2024"]

def safe_filename(text):
    """Windows-safe filename (removes |, Δ, θ, etc.)."""
    bad = ['<','>',';',':','"','/','\\','|','?','*','Δ','θ','→','–','—']
    for ch in bad:
        text = text.replace(ch, '_')
    text = " ".join(text.split())
    return text.strip().replace(" ", "_")

def raster_to_values(raster_path):
    """Read raster to 1D array and remove NoData."""
    ras = arcpy.Raster(raster_path)
    arr = arcpy.RasterToNumPyArray(ras, nodata_to_value=None).flatten()
    nd = ras.noDataValue
    if nd is not None:
        arr = arr[arr != nd]
    return arr

print("✅ Setup complete.")
print("Ten-year mean folder:", ten_year_dir)
print("Decadal change folder:", decadal_dir)
print("Multi-interval folder:", multi_dir)
print("Figure root:", fig_root)
```

```python
# =============================================================
# CELL 2 — TEN-YEAR MEAN DISTRIBUTION (PER DECADE)
# PURPOSE:
#   For each season and each decade (10-year mean raster),
#   plot the pixel value distribution (0–360°) as a LINE chart.
#   (Updated: season-based colors only)
# OUTPUT:
#   Saves PNGs into: 2_TenYear_Line_PerDecade
# =============================================================

# -------------------------------------------------------------
# Season colors (consistent across thesis)
# -------------------------------------------------------------
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red"
}


def plot_ten_year_distribution_line(raster_path, title, out_folder, season, bin_width=1):
    """Line chart of histogram counts for 0–360 mean-direction rasters."""
    if not arcpy.Exists(raster_path):
        print("⚠️ Missing:", raster_path)
        return

    values = raster_to_values(raster_path)

    bins = list(range(0, 361, bin_width))
    counts, edges = plt.hist(values, bins=bins)[0:2]
    plt.clf()

    centers = [(edges[i] + edges[i + 1]) / 2 for i in range(len(edges) - 1)]

    plt.figure(figsize=(10, 5))
    plt.plot(
        centers,
        counts,
        linewidth=2.5,
        color=season_colors.get(season)
    )

    plt.xlabel("Mean Wind Direction (°) [0–360]")
    plt.ylabel("Pixel Count")
    plt.title(title + f" (bin={bin_width}°)")

    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    out_png = os.path.join(
        out_folder,
        safe_filename(title + f"_bin{bin_width}") + ".png"
    )

    plt.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)


# -------------------------------------------------------------
# Run for all seasons and decades
# -------------------------------------------------------------
for s in seasons_list:
    for d in decades:
        rpath = os.path.join(ten_year_dir, f"{s}_10YearMean_{d}.tif")
        title = f"{s} Ten-Year Mean Direction Distribution ({d})"

        plot_ten_year_distribution_line(
            raster_path=rpath,
            title=title,
            out_folder=col2,
            season=s,
            bin_width=1
        )

print("\n🎯 Cell 2 complete — season-colored per-decade figures created.")
```

```python
# =============================================================
# CELL 5B-YEARLY — MEDIAN |Δθ| TREND (BASELINE = 1985–2024 CLIMATOLOGY) VS EVERY YEAR
#
# UPDATED (fixes your FileNotFoundError):
#   ✅ Ensures output folder (col5) exists before saving PNGs + CSV
#   ✅ Also creates YEARLY_CACHE folders
#   ✅ Everything else kept the same
#
# Assumes these exist from earlier cells:
#   base_dir, col5, seasons_list, safe_filename, raster_to_values
# =============================================================

import os, math, csv
import arcpy
from arcpy.sa import Cos, Sin, ATan2, Raster, Con
import matplotlib.pyplot as plt
import numpy as np

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

# --- INPUT folder (monthly rasters) ---
era5_dir = r"C:\Users\SHIDEBNA\Downloads\Thesis_Analysis\ERA5_RWD"

# --- Year range to plot ---
start_year = 1985
end_year = 2024

# --- Baseline range = full record ---
baseline_start = 1985
baseline_end   = 2024

# --- Consistent season colors ---
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red"
}

# -------------------------------------------------------------
# ✅ CRITICAL FIX: ensure output folder exists
# (this is what caused your FileNotFoundError)
# -------------------------------------------------------------
os.makedirs(col5, exist_ok=True)

# --- CACHE folder for yearly products ---
cache_root = os.path.join(base_dir, "YEARLY_CACHE")
cache_mean = os.path.join(cache_root, "yearly_season_mean")
cache_base = os.path.join(cache_root, "baseline_season_mean")
cache_diff = os.path.join(cache_root, "yearly_change")

for p in [cache_root, cache_mean, cache_base, cache_diff]:
    os.makedirs(p, exist_ok=True)

# --- Seasons definition (meteorological) ---
season_months = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11]
}

# -------------------------------------------------------------
# Circular mean for direction rasters (0–360)
# -------------------------------------------------------------
def vector_average(raster_list):
    n = len(raster_list)
    if n == 0:
        return None

    r0 = Raster(raster_list[0])
    cos_sum = Cos(r0 * math.pi / 180.0)
    sin_sum = Sin(r0 * math.pi / 180.0)

    for p in raster_list[1:]:
        r = Raster(p)
        cos_sum = cos_sum + Cos(r * math.pi / 180.0)
        sin_sum = sin_sum + Sin(r * math.pi / 180.0)

    cos_mean = cos_sum / float(n)
    sin_mean = sin_sum / float(n)

    mean_dir = ATan2(sin_mean, cos_mean) * 180.0 / math.pi
    mean_dir = Con(mean_dir < 0, mean_dir + 360, mean_dir)
    mean_dir = Con(mean_dir >= 360, mean_dir - 360, mean_dir)
    return mean_dir

# -------------------------------------------------------------
# Angular diff normalized to ±180  (anomaly relative to baseline)
# -------------------------------------------------------------
def angular_diff(r1, r2):
    raw = Raster(r2) - Raster(r1)
    adj = Con(raw > 180, raw - 360,
              Con(raw < -180, raw + 360, raw))
    return adj

# -------------------------------------------------------------
# Find monthly files for (year, month) allowing padded/nonpadded
# -------------------------------------------------------------
def find_month_file(year, month):
    p1 = os.path.join(era5_dir, f"RWD_Global_{year}_{month}.tif")
    p2 = os.path.join(era5_dir, f"RWD_Global_{year}_{month:02d}.tif")
    if os.path.exists(p1):
        return p1
    if os.path.exists(p2):
        return p2
    return None

# -------------------------------------------------------------
# Build seasonal raster list for a given (season, year)
# DJF(year) = Dec(year-1) + Jan(year) + Feb(year)
# -------------------------------------------------------------
def seasonal_files_for_year(season, year):
    months = season_months[season]
    files = []
    for m in months:
        y = year - 1 if m == 12 else year
        fp = find_month_file(y, m)
        if fp:
            files.append(fp)
    return files

# -------------------------------------------------------------
# Save yearly seasonal mean raster (cached)
# -------------------------------------------------------------
def make_yearly_season_mean(season, year):
    out_ras = os.path.join(cache_mean, f"{season}_YearMean_{year}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    rasters = seasonal_files_for_year(season, year)
    if len(rasters) < 3:
        print(f"⚠️ {season} {year}: missing months ({len(rasters)}/3). Skipped.")
        return None

    mean_ras = vector_average(rasters)
    if mean_ras is None:
        return None

    mean_ras.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Baseline = FULL-RECORD seasonal climatology (1985–2024)
# circular mean of yearly seasonal means.
# -------------------------------------------------------------
def make_baseline_mean(season):
    out_ras = os.path.join(cache_base, f"{season}_BaselineMean_{baseline_start}-{baseline_end}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    baseline_year_means = []
    for y in range(baseline_start, baseline_end + 1):
        yr = make_yearly_season_mean(season, y)
        if yr:
            baseline_year_means.append(yr)

    if not baseline_year_means:
        print(f"⚠️ Baseline missing for {season}")
        return None

    base_mean = vector_average(baseline_year_means)
    if base_mean is None:
        return None

    base_mean.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Compute baseline->year anomaly raster (cached)
# -------------------------------------------------------------
def make_yearly_anomaly(season, year, baseline_ras):
    out_ras = os.path.join(
        cache_diff,
        f"{season}_Anomaly_BaselineClim_{baseline_start}-{baseline_end}_to_{year}.tif"
    )
    if arcpy.Exists(out_ras):
        return out_ras

    yr_mean = make_yearly_season_mean(season, year)
    if not yr_mean:
        return None

    diff = angular_diff(baseline_ras, yr_mean)
    diff.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Median |anomaly| from raster (uses your raster_to_values helper)
# -------------------------------------------------------------
def median_abs_anomaly_from_raster(raster_path):
    vals = raster_to_values(raster_path)  # must exist from earlier cells
    if len(vals) == 0:
        return None

    abs_vals = np.abs(np.array(vals, dtype=float))
    abs_vals = abs_vals[np.isfinite(abs_vals)]
    if abs_vals.size == 0:
        return None

    return float(np.median(abs_vals))

# -------------------------------------------------------------
# Plot helper: yearly + mean line + mean label (BLACK, not bold)
# -------------------------------------------------------------
def plot_yearly_and_mean(ax, x, y, color, season_label, mean_label_pos="top"):
    x = np.array(x, dtype=int)
    y = np.array(y, dtype=float)

    ax.plot(
        x, y,
        marker="o",
        markersize=5,
        linewidth=1.2,
        alpha=0.45,
        color=color,
        label=f"{season_label} yearly",
        zorder=2
    )

    mean_val = float(np.nanmean(y))
    ax.axhline(
        mean_val,
        linestyle="--",
        linewidth=3.0,
        alpha=0.95,
        color=color,
        label=f"{season_label} mean",
        zorder=1
    )

    txt = f"{season_label} mean = {mean_val:.3f}°"
    if mean_label_pos == "top":
        ax.text(0.98, 0.92, txt, transform=ax.transAxes,
                ha="right", va="top", color="black", fontsize=11, fontweight="normal")
    else:
        ax.text(0.98, 0.08, txt, transform=ax.transAxes,
                ha="right", va="bottom", color="black", fontsize=11, fontweight="normal")

    return mean_val

# -------------------------------------------------------------
# FINAL TABLE EXPORT (CSV)
# -------------------------------------------------------------
def export_final_table(yearly_dict, season_means, out_csv):
    os.makedirs(os.path.dirname(out_csv), exist_ok=True)  # ✅ ensure folder exists
    years = list(range(start_year, end_year + 1))
    seasons = list(seasons_list)

    with open(out_csv, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)

        w.writerow(["Year"] + seasons)

        for y in years:
            row = [y]
            for s in seasons:
                v = yearly_dict.get(s, {}).get(y, "")
                row.append("" if v is None else f"{v:.6f}")
            w.writerow(row)

        w.writerow([])
        w.writerow(["MEAN"] + [f"{season_means.get(s, float('nan')):.6f}" for s in seasons])

        w.writerow([])
        w.writerow(["Baseline", f"{baseline_start}-{baseline_end}"])
        w.writerow(["Metric", "Median(|Anomaly|) per year (degrees)"])

    print("✅ Saved table:", out_csv)

# =============================================================
# PART A — Individual season plots
# =============================================================
def plot_yearly_trend_for_season(season):
    baseline_ras = make_baseline_mean(season)
    if not baseline_ras:
        return None, None

    years = []
    medians = []

    for y in range(start_year, end_year + 1):
        anom_ras = make_yearly_anomaly(season, y, baseline_ras)
        if not anom_ras:
            continue
        m = median_abs_anomaly_from_raster(anom_ras)
        if m is None:
            continue
        years.append(y)
        medians.append(m)

    if not medians:
        print(f"⚠️ No yearly values for {season}")
        return None, None

    fig, ax = plt.subplots(figsize=(11, 5))

    mean_val = plot_yearly_and_mean(
        ax, years, medians,
        season_colors.get(season),
        season_label=season,
        mean_label_pos="top"
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"{season} Yearly Median |Anomalies| vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend()
    fig.tight_layout()

    # ✅ ensure output folder exists before save
    os.makedirs(col5, exist_ok=True)

    out_png = os.path.join(
        col5,
        safe_filename(f"{season}_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

    return dict(zip(years, medians)), mean_val

# =============================================================
# PART B — Combined all-seasons plot
# =============================================================
def plot_yearly_trend_all_seasons(yearly_dict, season_means):
    fig, ax = plt.subplots(figsize=(11, 5))

    for season in seasons_list:
        season_years = sorted(yearly_dict.get(season, {}).keys())
        if not season_years:
            continue
        med = [yearly_dict[season][y] for y in season_years]

        ax.plot(
            season_years, med,
            marker="o",
            markersize=5,
            linewidth=1.2,
            alpha=0.45,
            color=season_colors.get(season),
            label=f"{season} yearly",
            zorder=2
        )

        mean_val = season_means.get(season, None)
        if mean_val is not None and np.isfinite(mean_val):
            ax.axhline(
                mean_val,
                linestyle="--",
                linewidth=3.0,
                alpha=0.95,
                color=season_colors.get(season),
                label=f"{season} mean",
                zorder=1
            )

    lines = []
    for s in seasons_list:
        mv = season_means.get(s, np.nan)
        if np.isfinite(mv):
            lines.append(f"{s} mean = {mv:.3f}°")
    ax.text(
        0.02, 0.98,
        "\n".join(lines),
        transform=ax.transAxes,
        ha="left", va="top",
        fontsize=11,
        fontweight="normal",
        color="black",
        bbox=dict(boxstyle="round", alpha=0.12)
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"Yearly Median |Anomalies| by Season vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend(title="Season", ncols=2)
    fig.tight_layout()

    # ✅ ensure output folder exists before save
    os.makedirs(col5, exist_ok=True)

    out_png = os.path.join(
        col5,
        safe_filename(f"ALL_Seasons_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

# -------------------------------------------------------------
# RUN + EXPORT TABLE
# -------------------------------------------------------------
yearly_table = {}     # season -> {year: median(|anomaly|)}
season_means = {}     # season -> mean

for s in seasons_list:
    yd, mv = plot_yearly_trend_for_season(s)
    if yd is not None:
        yearly_table[s] = yd
    if mv is not None:
        season_means[s] = mv

plot_yearly_trend_all_seasons(yearly_table, season_means)

out_csv = os.path.join(
    col5,
    safe_filename(f"FINAL_TABLE_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".csv"
)
export_final_table(yearly_table, season_means, out_csv)

print("\n🎯 DONE: Baseline is the FULL-RECORD seasonal climatology =", f"{baseline_start}-{baseline_end}")
print("   Cached rasters stored in:", cache_root)
print("   Table saved to:", out_csv)
```

```python
# =============================================================
# CELL 5 — MEDIAN |Δθ| TREND (EARLIEST BASELINE ONLY)
# PURPOSE:
#   (1) Produce ONE trend figure per season using baseline 1985–1994
#   (2) Produce ONE combined figure showing all seasons together
#   X-axis shows DECADE NAMES ONLY.
# OUTPUT:
#   Saves PNGs into: 5_MedianAbsTrend_ByBaselineDecade
# =============================================================

# -------------------------------------------------------------
# Consistent season colors (used everywhere)
# -------------------------------------------------------------
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red",
}


def median_abs_change(raster_path):
    values = raster_to_values(raster_path)
    if len(values) == 0:
        return None

    vals_abs = [abs(float(v)) for v in values]
    vals_abs.sort()
    n = len(vals_abs)
    mid = n // 2

    return vals_abs[mid] if n % 2 == 1 else (vals_abs[mid-1] + vals_abs[mid]) / 2


# -------------------------------------------------------------
# PART A — Individual season trend figures
# -------------------------------------------------------------
def plot_trend_earliest_to_latest(season):

    baseline = decades[0]     # 1985–1994
    later = decades[1:]       # later decades

    rasters = []
    xlabels = []

    for d2 in later:
        rasters.append(
            os.path.join(multi_dir, f"{season}_Change_{baseline}_vs_{d2}.tif")
        )
        xlabels.append(d2)

    medians = []
    used_labels = []

    for rp, lab in zip(rasters, xlabels):
        if not arcpy.Exists(rp):
            print("⚠️ Missing:", rp)
            continue

        m = median_abs_change(rp)
        if m is not None:
            medians.append(m)
            used_labels.append(lab)

    if not medians:
        print(f"⚠️ No trend data for {season}")
        return

    plt.figure(figsize=(9, 5))

    plt.plot(
        used_labels,
        medians,
        marker="o",
        linewidth=2,
        color=season_colors.get(season),   # ✅ SAME COLOR
        label=season
    )

    plt.xlabel("Comparison Decade")
    plt.ylabel("Median Absolute Directional Change (°)")
    plt.title(f"{season} Median Absolute Directional Change (Baseline: {baseline})")

    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()

    out_png = os.path.join(
        col5,
        safe_filename(f"{season}_MedianAbsTrend_Baseline_{baseline}") + ".png"
    )

    plt.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)


# -------------------------------------------------------------
# PART B — Combined all-seasons trend figure
# -------------------------------------------------------------
def combined_season_trend_baseline_earliest():

    baseline = decades[0]
    later = decades[1:]

    x = list(range(len(later)))
    xlabels = later

    plt.figure(figsize=(10, 5))

    for s in seasons_list:

        medians = []

        for d2 in later:
            rp = os.path.join(multi_dir, f"{s}_Change_{baseline}_vs_{d2}.tif")

            if arcpy.Exists(rp):
                medians.append(median_abs_change(rp))
            else:
                medians.append(None)

        xx = [x[i] for i in range(len(medians)) if medians[i] is not None]
        yy = [medians[i] for i in range(len(medians)) if medians[i] is not None]

        if yy:
            plt.plot(
                xx,
                yy,
                marker="o",
                linewidth=2,
                color=season_colors.get(s),   # ✅ SAME COLORS
                label=s
            )

    plt.xticks(x, xlabels)
    plt.xlabel("Comparison Decade")
    plt.ylabel("Median Absolute Directional Change (°)")
    plt.title(f"Median Absolute Directional Change by Season (Baseline: {baseline})")

    plt.grid(True, alpha=0.3)
    plt.legend(title="Season")
    plt.tight_layout()

    out_png = os.path.join(
        col5,
        safe_filename(f"ALL_Seasons_MedianAbsTrend_Baseline_{baseline}") + ".png"
    )

    plt.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)


# -------------------------------------------------------------
# RUN BOTH
# -------------------------------------------------------------
for s in seasons_list:
    plot_trend_earliest_to_latest(s)

combined_season_trend_baseline_earliest()

print("\n🎯 CELL 5 COMPLETE — colors now fully consistent.")
```

```python
# =============================================================
# CELL 5B-YEARLY — MEDIAN |Δθ| TREND (BASELINE = 1985–2024 CLIMATOLOGY) VS EVERY YEAR
#
# UPDATED (per your request):
#   ✅ Replace "Directional Change" wording with "Anomalies" in labels/titles
#   ✅ Mean value text is BLACK and NOT bold
#   ✅ Keeps your styling: yearly thin+markers + mean thick dashed (season color)
#   ✅ Exports final CSV table
# =============================================================

import os, math
import arcpy
from arcpy.sa import Cos, Sin, ATan2, Raster, Con
import matplotlib.pyplot as plt
import numpy as np
import csv

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

# --- INPUT folder (monthly rasters) ---
era5_dir = r"C:\Users\SHIDEBNA\Downloads\Thesis_Analysis\ERA5_RWD"

# --- CACHE folder for yearly products ---
cache_root = os.path.join(base_dir, "YEARLY_CACHE")
cache_mean = os.path.join(cache_root, "yearly_season_mean")
cache_base = os.path.join(cache_root, "baseline_season_mean")
cache_diff = os.path.join(cache_root, "yearly_change")

for p in [cache_root, cache_mean, cache_base, cache_diff]:
    os.makedirs(p, exist_ok=True)

# --- Seasons definition (meteorological) ---
season_months = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11]
}

# --- Consistent season colors ---
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red"
}

# --- Year range to plot ---
start_year = 1985
end_year = 2024

# --- Baseline range = full record ---
baseline_start = 1985
baseline_end   = 2024


# -------------------------------------------------------------
# Circular mean for direction rasters (0–360)
# -------------------------------------------------------------
def vector_average(raster_list):
    n = len(raster_list)
    if n == 0:
        return None

    r0 = Raster(raster_list[0])
    cos_sum = Cos(r0 * math.pi / 180.0)
    sin_sum = Sin(r0 * math.pi / 180.0)

    for p in raster_list[1:]:
        r = Raster(p)
        cos_sum = cos_sum + Cos(r * math.pi / 180.0)
        sin_sum = sin_sum + Sin(r * math.pi / 180.0)

    cos_mean = cos_sum / float(n)
    sin_mean = sin_sum / float(n)

    mean_dir = ATan2(sin_mean, cos_mean) * 180.0 / math.pi
    mean_dir = Con(mean_dir < 0, mean_dir + 360, mean_dir)
    mean_dir = Con(mean_dir >= 360, mean_dir - 360, mean_dir)
    return mean_dir


# -------------------------------------------------------------
# Angular diff normalized to ±180
# (This is your "anomaly" relative to baseline, in degrees)
# -------------------------------------------------------------
def angular_diff(r1, r2):
    raw = Raster(r2) - Raster(r1)
    adj = Con(raw > 180, raw - 360,
              Con(raw < -180, raw + 360, raw))
    return adj


# -------------------------------------------------------------
# Find monthly files for (year, month) allowing padded/nonpadded
# -------------------------------------------------------------
def find_month_file(year, month):
    p1 = os.path.join(era5_dir, f"RWD_Global_{year}_{month}.tif")
    p2 = os.path.join(era5_dir, f"RWD_Global_{year}_{month:02d}.tif")
    if os.path.exists(p1):
        return p1
    if os.path.exists(p2):
        return p2
    return None


# -------------------------------------------------------------
# Build seasonal raster list for a given (season, year)
# DJF(year) = Dec(year-1) + Jan(year) + Feb(year)
# -------------------------------------------------------------
def seasonal_files_for_year(season, year):
    months = season_months[season]
    files = []
    for m in months:
        y = year - 1 if m == 12 else year
        fp = find_month_file(y, m)
        if fp:
            files.append(fp)
    return files


# -------------------------------------------------------------
# Save yearly seasonal mean raster (cached)
# -------------------------------------------------------------
def make_yearly_season_mean(season, year):
    out_ras = os.path.join(cache_mean, f"{season}_YearMean_{year}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    rasters = seasonal_files_for_year(season, year)
    if len(rasters) < 3:
        print(f"⚠️ {season} {year}: missing months ({len(rasters)}/3). Skipped.")
        return None

    mean_ras = vector_average(rasters)
    if mean_ras is None:
        return None

    mean_ras.save(out_ras)
    return out_ras


# -------------------------------------------------------------
# Baseline = FULL-RECORD seasonal climatology (1985–2024)
#   circular mean of yearly seasonal means.
# -------------------------------------------------------------
def make_baseline_mean(season):
    out_ras = os.path.join(cache_base, f"{season}_BaselineMean_{baseline_start}-{baseline_end}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    baseline_year_means = []
    for y in range(baseline_start, baseline_end + 1):
        yr = make_yearly_season_mean(season, y)
        if yr:
            baseline_year_means.append(yr)

    if not baseline_year_means:
        print(f"⚠️ Baseline missing for {season}")
        return None

    base_mean = vector_average(baseline_year_means)
    if base_mean is None:
        return None

    base_mean.save(out_ras)
    return out_ras


# -------------------------------------------------------------
# Compute baseline->year anomaly raster (cached)
# -------------------------------------------------------------
def make_yearly_anomaly(season, year, baseline_ras):
    out_ras = os.path.join(
        cache_diff,
        f"{season}_Anomaly_BaselineClim_{baseline_start}-{baseline_end}_to_{year}.tif"
    )
    if arcpy.Exists(out_ras):
        return out_ras

    yr_mean = make_yearly_season_mean(season, year)
    if not yr_mean:
        return None

    diff = angular_diff(baseline_ras, yr_mean)
    diff.save(out_ras)
    return out_ras


# -------------------------------------------------------------
# Median |anomaly| from raster (uses your raster_to_values helper)
# -------------------------------------------------------------
def median_abs_anomaly_from_raster(raster_path):
    vals = raster_to_values(raster_path)  # must exist from earlier cells
    if len(vals) == 0:
        return None

    abs_vals = np.abs(np.array(vals, dtype=float))
    abs_vals = abs_vals[np.isfinite(abs_vals)]
    if abs_vals.size == 0:
        return None

    return float(np.median(abs_vals))


# =============================================================
# Plot helper: yearly + mean line + mean label (BLACK, not bold)
# =============================================================
def plot_yearly_and_mean(ax, x, y, color, season_label, mean_label_pos="top"):
    x = np.array(x, dtype=int)
    y = np.array(y, dtype=float)

    # Yearly: thin + markers
    ax.plot(
        x, y,
        marker="o",
        markersize=5,
        linewidth=1.2,
        alpha=0.45,
        color=color,
        label=f"{season_label} yearly",
        zorder=2
    )

    # Mean: thick dashed (same season color)
    mean_val = float(np.nanmean(y))
    ax.axhline(
        mean_val,
        linestyle="--",
        linewidth=3.0,
        alpha=0.95,
        color=color,
        label=f"{season_label} mean",
        zorder=1
    )

    # Mean label in BLACK, not bold
    txt = f"{season_label} mean = {mean_val:.3f}°"
    if mean_label_pos == "top":
        ax.text(
            0.98, 0.92,
            txt,
            transform=ax.transAxes,
            ha="right", va="top",
            color="black",
            fontsize=11,
            fontweight="normal"
        )
    else:
        ax.text(
            0.98, 0.08,
            txt,
            transform=ax.transAxes,
            ha="right", va="bottom",
            color="black",
            fontsize=11,
            fontweight="normal"
        )

    return mean_val


# =============================================================
# FINAL TABLE EXPORT (CSV)
# =============================================================
def export_final_table(yearly_dict, season_means, out_csv):
    years = list(range(start_year, end_year + 1))
    seasons = list(seasons_list)

    with open(out_csv, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)

        w.writerow(["Year"] + seasons)

        for y in years:
            row = [y]
            for s in seasons:
                v = yearly_dict.get(s, {}).get(y, "")
                row.append("" if v is None else f"{v:.6f}")
            w.writerow(row)

        w.writerow([])
        w.writerow(["MEAN"] + [f"{season_means.get(s, float('nan')):.6f}" for s in seasons])

        w.writerow([])
        w.writerow(["Baseline", f"{baseline_start}-{baseline_end}"])
        w.writerow(["Metric", "Median(|Anomaly|) per year (degrees)"])

    print("✅ Saved table:", out_csv)


# =============================================================
# PART A — Individual season yearly anomaly plots
# =============================================================
def plot_yearly_trend_for_season(season):
    baseline_ras = make_baseline_mean(season)
    if not baseline_ras:
        return None, None

    years = []
    medians = []

    for y in range(start_year, end_year + 1):
        anom_ras = make_yearly_anomaly(season, y, baseline_ras)
        if not anom_ras:
            continue
        m = median_abs_anomaly_from_raster(anom_ras)
        if m is None:
            continue
        years.append(y)
        medians.append(m)

    if not medians:
        print(f"⚠️ No yearly values for {season}")
        return None, None

    fig, ax = plt.subplots(figsize=(11, 5))

    mean_val = plot_yearly_and_mean(
        ax,
        years,
        medians,
        season_colors.get(season),
        season_label=season,
        mean_label_pos="top"
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"{season} Yearly Median |Anomalies| vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend()
    fig.tight_layout()

    out_png = os.path.join(
        col5,
        safe_filename(f"{season}_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

    return dict(zip(years, medians)), mean_val


# =============================================================
# PART B — Combined all-seasons yearly anomaly plot
# =============================================================
def plot_yearly_trend_all_seasons(yearly_dict, season_means):
    fig, ax = plt.subplots(figsize=(11, 5))

    for season in seasons_list:
        season_years = sorted(yearly_dict.get(season, {}).keys())
        if not season_years:
            continue
        med = [yearly_dict[season][y] for y in season_years]

        ax.plot(
            season_years, med,
            marker="o",
            markersize=5,
            linewidth=1.2,
            alpha=0.45,
            color=season_colors.get(season),
            label=f"{season} yearly",
            zorder=2
        )

        mean_val = season_means.get(season, None)
        if mean_val is not None and np.isfinite(mean_val):
            ax.axhline(
                mean_val,
                linestyle="--",
                linewidth=3.0,
                alpha=0.95,
                color=season_colors.get(season),
                label=f"{season} mean",
                zorder=1
            )

    # Mean-value text block in BLACK, not bold
    lines = []
    for s in seasons_list:
        mv = season_means.get(s, np.nan)
        if np.isfinite(mv):
            lines.append(f"{s} mean = {mv:.3f}°")
    ax.text(
        0.02, 0.98,
        "\n".join(lines),
        transform=ax.transAxes,
        ha="left", va="top",
        fontsize=11,
        fontweight="normal",
        color="black",
        bbox=dict(boxstyle="round", alpha=0.12)
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"Yearly Median |Anomalies| by Season vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend(title="Season", ncols=2)
    fig.tight_layout()

    out_png = os.path.join(
        col5,
        safe_filename(f"ALL_Seasons_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)


# -------------------------------------------------------------
# RUN + COLLECT TABLE DATA
# -------------------------------------------------------------
yearly_table = {}     # season -> {year: median(|anomaly|)}
season_means = {}     # season -> mean

for s in seasons_list:
    yd, mv = plot_yearly_trend_for_season(s)
    if yd is not None:
        yearly_table[s] = yd
    if mv is not None:
        season_means[s] = mv

plot_yearly_trend_all_seasons(yearly_table, season_means)

# Export final table (CSV) into same output folder as plots
out_csv = os.path.join(
    col5,
    safe_filename(f"FINAL_TABLE_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".csv"
)
export_final_table(yearly_table, season_means, out_csv)

print("\n🎯 DONE: Baseline is the FULL-RECORD seasonal climatology =", f"{baseline_start}-{baseline_end}")
print("   Cached rasters stored in:", cache_root)
print("   Table saved to:", out_csv)
```

```python
# =============================================================
# CELL 5B-YEARLY — MEDIAN |Δθ| TREND (BASELINE = 1985–2024 CLIMATOLOGY) VS EVERY YEAR
#
# UPDATED (fixes your FileNotFoundError):
#   ✅ Ensures output folder (col5) exists before saving PNGs + CSV
#   ✅ Also creates YEARLY_CACHE folders
#   ✅ Everything else kept the same
#
# Assumes these exist from earlier cells:
#   base_dir, col5, seasons_list, safe_filename, raster_to_values
# =============================================================

import os, math, csv
import arcpy
from arcpy.sa import Cos, Sin, ATan2, Raster, Con
import matplotlib.pyplot as plt
import numpy as np

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

# --- INPUT folder (monthly rasters) ---
era5_dir = r"C:\Users\SHIDEBNA\Downloads\Thesis_Analysis\ERA5_RWD"

# --- Year range to plot ---
start_year = 1985
end_year = 2024

# --- Baseline range = full record ---
baseline_start = 1985
baseline_end   = 2024

# --- Consistent season colors ---
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red"
}

# -------------------------------------------------------------
# ✅ CRITICAL FIX: ensure output folder exists
# (this is what caused your FileNotFoundError)
# -------------------------------------------------------------
os.makedirs(col5, exist_ok=True)

# --- CACHE folder for yearly products ---
cache_root = os.path.join(base_dir, "YEARLY_CACHE")
cache_mean = os.path.join(cache_root, "yearly_season_mean")
cache_base = os.path.join(cache_root, "baseline_season_mean")
cache_diff = os.path.join(cache_root, "yearly_change")

for p in [cache_root, cache_mean, cache_base, cache_diff]:
    os.makedirs(p, exist_ok=True)

# --- Seasons definition (meteorological) ---
season_months = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11]
}

# -------------------------------------------------------------
# Circular mean for direction rasters (0–360)
# -------------------------------------------------------------
def vector_average(raster_list):
    n = len(raster_list)
    if n == 0:
        return None

    r0 = Raster(raster_list[0])
    cos_sum = Cos(r0 * math.pi / 180.0)
    sin_sum = Sin(r0 * math.pi / 180.0)

    for p in raster_list[1:]:
        r = Raster(p)
        cos_sum = cos_sum + Cos(r * math.pi / 180.0)
        sin_sum = sin_sum + Sin(r * math.pi / 180.0)

    cos_mean = cos_sum / float(n)
    sin_mean = sin_sum / float(n)

    mean_dir = ATan2(sin_mean, cos_mean) * 180.0 / math.pi
    mean_dir = Con(mean_dir < 0, mean_dir + 360, mean_dir)
    mean_dir = Con(mean_dir >= 360, mean_dir - 360, mean_dir)
    return mean_dir

# -------------------------------------------------------------
# Angular diff normalized to ±180  (anomaly relative to baseline)
# -------------------------------------------------------------
def angular_diff(r1, r2):
    raw = Raster(r2) - Raster(r1)
    adj = Con(raw > 180, raw - 360,
              Con(raw < -180, raw + 360, raw))
    return adj

# -------------------------------------------------------------
# Find monthly files for (year, month) allowing padded/nonpadded
# -------------------------------------------------------------
def find_month_file(year, month):
    p1 = os.path.join(era5_dir, f"RWD_Global_{year}_{month}.tif")
    p2 = os.path.join(era5_dir, f"RWD_Global_{year}_{month:02d}.tif")
    if os.path.exists(p1):
        return p1
    if os.path.exists(p2):
        return p2
    return None

# -------------------------------------------------------------
# Build seasonal raster list for a given (season, year)
# DJF(year) = Dec(year-1) + Jan(year) + Feb(year)
# -------------------------------------------------------------
def seasonal_files_for_year(season, year):
    months = season_months[season]
    files = []
    for m in months:
        y = year - 1 if m == 12 else year
        fp = find_month_file(y, m)
        if fp:
            files.append(fp)
    return files

# -------------------------------------------------------------
# Save yearly seasonal mean raster (cached)
# -------------------------------------------------------------
def make_yearly_season_mean(season, year):
    out_ras = os.path.join(cache_mean, f"{season}_YearMean_{year}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    rasters = seasonal_files_for_year(season, year)
    if len(rasters) < 3:
        print(f"⚠️ {season} {year}: missing months ({len(rasters)}/3). Skipped.")
        return None

    mean_ras = vector_average(rasters)
    if mean_ras is None:
        return None

    mean_ras.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Baseline = FULL-RECORD seasonal climatology (1985–2024)
# circular mean of yearly seasonal means.
# -------------------------------------------------------------
def make_baseline_mean(season):
    out_ras = os.path.join(cache_base, f"{season}_BaselineMean_{baseline_start}-{baseline_end}.tif")
    if arcpy.Exists(out_ras):
        return out_ras

    baseline_year_means = []
    for y in range(baseline_start, baseline_end + 1):
        yr = make_yearly_season_mean(season, y)
        if yr:
            baseline_year_means.append(yr)

    if not baseline_year_means:
        print(f"⚠️ Baseline missing for {season}")
        return None

    base_mean = vector_average(baseline_year_means)
    if base_mean is None:
        return None

    base_mean.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Compute baseline->year anomaly raster (cached)
# -------------------------------------------------------------
def make_yearly_anomaly(season, year, baseline_ras):
    out_ras = os.path.join(
        cache_diff,
        f"{season}_Anomaly_BaselineClim_{baseline_start}-{baseline_end}_to_{year}.tif"
    )
    if arcpy.Exists(out_ras):
        return out_ras

    yr_mean = make_yearly_season_mean(season, year)
    if not yr_mean:
        return None

    diff = angular_diff(baseline_ras, yr_mean)
    diff.save(out_ras)
    return out_ras

# -------------------------------------------------------------
# Median |anomaly| from raster (uses your raster_to_values helper)
# -------------------------------------------------------------
def median_abs_anomaly_from_raster(raster_path):
    vals = raster_to_values(raster_path)  # must exist from earlier cells
    if len(vals) == 0:
        return None

    abs_vals = np.abs(np.array(vals, dtype=float))
    abs_vals = abs_vals[np.isfinite(abs_vals)]
    if abs_vals.size == 0:
        return None

    return float(np.median(abs_vals))

# -------------------------------------------------------------
# Plot helper: yearly + mean line + mean label (BLACK, not bold)
# -------------------------------------------------------------
def plot_yearly_and_mean(ax, x, y, color, season_label, mean_label_pos="top"):
    x = np.array(x, dtype=int)
    y = np.array(y, dtype=float)

    ax.plot(
        x, y,
        marker="o",
        markersize=5,
        linewidth=1.2,
        alpha=0.45,
        color=color,
        label=f"{season_label} yearly",
        zorder=2
    )

    mean_val = float(np.nanmean(y))
    ax.axhline(
        mean_val,
        linestyle="--",
        linewidth=3.0,
        alpha=0.95,
        color=color,
        label=f"{season_label} mean",
        zorder=1
    )

    txt = f"{season_label} mean = {mean_val:.3f}°"
    if mean_label_pos == "top":
        ax.text(0.98, 0.92, txt, transform=ax.transAxes,
                ha="right", va="top", color="black", fontsize=11, fontweight="normal")
    else:
        ax.text(0.98, 0.08, txt, transform=ax.transAxes,
                ha="right", va="bottom", color="black", fontsize=11, fontweight="normal")

    return mean_val

# -------------------------------------------------------------
# FINAL TABLE EXPORT (CSV)
# -------------------------------------------------------------
def export_final_table(yearly_dict, season_means, out_csv):
    os.makedirs(os.path.dirname(out_csv), exist_ok=True)  # ✅ ensure folder exists
    years = list(range(start_year, end_year + 1))
    seasons = list(seasons_list)

    with open(out_csv, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)

        w.writerow(["Year"] + seasons)

        for y in years:
            row = [y]
            for s in seasons:
                v = yearly_dict.get(s, {}).get(y, "")
                row.append("" if v is None else f"{v:.6f}")
            w.writerow(row)

        w.writerow([])
        w.writerow(["MEAN"] + [f"{season_means.get(s, float('nan')):.6f}" for s in seasons])

        w.writerow([])
        w.writerow(["Baseline", f"{baseline_start}-{baseline_end}"])
        w.writerow(["Metric", "Median(|Anomaly|) per year (degrees)"])

    print("✅ Saved table:", out_csv)

# =============================================================
# PART A — Individual season plots
# =============================================================
def plot_yearly_trend_for_season(season):
    baseline_ras = make_baseline_mean(season)
    if not baseline_ras:
        return None, None

    years = []
    medians = []

    for y in range(start_year, end_year + 1):
        anom_ras = make_yearly_anomaly(season, y, baseline_ras)
        if not anom_ras:
            continue
        m = median_abs_anomaly_from_raster(anom_ras)
        if m is None:
            continue
        years.append(y)
        medians.append(m)

    if not medians:
        print(f"⚠️ No yearly values for {season}")
        return None, None

    fig, ax = plt.subplots(figsize=(11, 5))

    mean_val = plot_yearly_and_mean(
        ax, years, medians,
        season_colors.get(season),
        season_label=season,
        mean_label_pos="top"
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"{season} Yearly Median |Anomalies| vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend()
    fig.tight_layout()

    # ✅ ensure output folder exists before save
    os.makedirs(col5, exist_ok=True)

    out_png = os.path.join(
        col5,
        safe_filename(f"{season}_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

    return dict(zip(years, medians)), mean_val

# =============================================================
# PART B — Combined all-seasons plot
# =============================================================
def plot_yearly_trend_all_seasons(yearly_dict, season_means):
    fig, ax = plt.subplots(figsize=(11, 5))

    for season in seasons_list:
        season_years = sorted(yearly_dict.get(season, {}).keys())
        if not season_years:
            continue
        med = [yearly_dict[season][y] for y in season_years]

        ax.plot(
            season_years, med,
            marker="o",
            markersize=5,
            linewidth=1.2,
            alpha=0.45,
            color=season_colors.get(season),
            label=f"{season} yearly",
            zorder=2
        )

        mean_val = season_means.get(season, None)
        if mean_val is not None and np.isfinite(mean_val):
            ax.axhline(
                mean_val,
                linestyle="--",
                linewidth=3.0,
                alpha=0.95,
                color=season_colors.get(season),
                label=f"{season} mean",
                zorder=1
            )

    lines = []
    for s in seasons_list:
        mv = season_means.get(s, np.nan)
        if np.isfinite(mv):
            lines.append(f"{s} mean = {mv:.3f}°")
    ax.text(
        0.02, 0.98,
        "\n".join(lines),
        transform=ax.transAxes,
        ha="left", va="top",
        fontsize=11,
        fontweight="normal",
        color="black",
        bbox=dict(boxstyle="round", alpha=0.12)
    )

    ax.set_xlabel("Year (seasonal means)")
    ax.set_ylabel("Median Absolute Anomalies (°)")
    ax.set_title(f"Yearly Median |Anomalies| by Season vs Baseline Climatology {baseline_start}-{baseline_end}")
    ax.grid(True, alpha=0.25)
    ax.legend(title="Season", ncols=2)
    fig.tight_layout()

    # ✅ ensure output folder exists before save
    os.makedirs(col5, exist_ok=True)

    out_png = os.path.join(
        col5,
        safe_filename(f"ALL_Seasons_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".png"
    )
    fig.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

# -------------------------------------------------------------
# RUN + EXPORT TABLE
# -------------------------------------------------------------
yearly_table = {}     # season -> {year: median(|anomaly|)}
season_means = {}     # season -> mean

for s in seasons_list:
    yd, mv = plot_yearly_trend_for_season(s)
    if yd is not None:
        yearly_table[s] = yd
    if mv is not None:
        season_means[s] = mv

plot_yearly_trend_all_seasons(yearly_table, season_means)

out_csv = os.path.join(
    col5,
    safe_filename(f"FINAL_TABLE_Yearly_MedianAbsAnomalies_BaselineClim_{baseline_start}-{baseline_end}") + ".csv"
)
export_final_table(yearly_table, season_means, out_csv)

print("\n🎯 DONE: Baseline is the FULL-RECORD seasonal climatology =", f"{baseline_start}-{baseline_end}")
print("   Cached rasters stored in:", cache_root)
print("   Table saved to:", out_csv)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# Consistent season colors (thesis-wide)
# -------------------------------------------------------------
season_colors = {
    "DJF": "tab:blue",
    "MAM": "tab:orange",
    "JJA": "tab:green",
    "SON": "tab:red",
}

# -------------------------------------------------------------
# Table storage
# -------------------------------------------------------------
summary_rows = []

def plot_change_continuous(raster_path, title, out_folder, season, d1, d2, bin_width=1):
    """
    Continuous histogram (-180..180) with frequency line overlay,
    plus median + IQR summary lines (thin), and writes stats to a table list.
    """
    
    if not arcpy.Exists(raster_path):
        print("⚠️ Missing:", raster_path)
        return
    
    values = raster_to_values(raster_path)
    
    # Safe empty check for list or numpy array
    if values is None or len(values) == 0:
        print("⚠️ Empty raster values:", raster_path)
        return
    
    values = np.array(values, dtype=float)
    
    # ---------------------------------------------------------
    # Stats
    # ---------------------------------------------------------
    median = np.percentile(values, 50)
    q1 = np.percentile(values, 25)
    q3 = np.percentile(values, 75)
    iqr = q3 - q1
    mean = np.mean(values)
    std = np.std(values)
    n = int(values.size)
    
    # Add to summary table
    summary_rows.append({
        "Season": season,
        "From": d1,
        "To": d2,
        "Median_deg": round(float(median), 2),
        "Q1_deg": round(float(q1), 2),
        "Q3_deg": round(float(q3), 2),
        "IQR_deg": round(float(iqr), 2),
        "Mean_deg": round(float(mean), 2),
        "Std_deg": round(float(std), 2),
        "Pixel_Count": n
    })
    
    # ---------------------------------------------------------
    # Plot
    # ---------------------------------------------------------
    bins = list(range(-180, 181, bin_width))
    color = season_colors.get(season, "gray")
    
    plt.figure(figsize=(10, 5))
    
    # Histogram
    counts, edges, _ = plt.hist(
        values,
        bins=bins,
        color=color,
        edgecolor="black",
        alpha=0.55
    )
    
    # Frequency line
    centers = [(edges[i] + edges[i + 1]) / 2 for i in range(len(edges) - 1)]
    plt.plot(centers, counts, color=color, linewidth=2.2)
    
    # Median + IQR lines (half thickness from earlier version)
    plt.axvline(median, color="black", linewidth=1.25, label="Median")
    plt.axvline(q1, color="black", linestyle="--", linewidth=0.9, label="IQR (25–75%)")
    plt.axvline(q3, color="black", linestyle="--", linewidth=0.9)
    
    # Zero reference
    plt.axvline(0, color="black", linewidth=0.6, alpha=0.6)
    
    # X ticks every 30°
    plt.xticks(range(-180, 181, 30))
    
    plt.xlabel("Directional Change (°) normalized to ±180")
    plt.ylabel("Pixel Count")
    # Updated title to exclude N (number of pixels)
    plt.title(f"{title}\nMedian={median:.2f}°, IQR={iqr:.2f}°")
    
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    
    out_png = os.path.join(out_folder, safe_filename(title + f"_continuous_bin{bin_width}") + ".png")
    plt.savefig(out_png, dpi=300)
    plt.show()
    print("✅ Saved:", out_png)

# -------------------------------------------------------------
# Define change comparisons (consecutive decades + long-term)
# -------------------------------------------------------------
pairs = [(decades[i], decades[i + 1]) for i in range(len(decades) - 1)] 
pairs.append((decades[0], decades[-1]))  # long-term (~40-year climatological span)

# -------------------------------------------------------------
# Generate figures
# -------------------------------------------------------------
for s in seasons_list:
    for d1, d2 in pairs:
        rpath = os.path.join(decadal_dir, f"{s}_Change_{d1}_vs_{d2}.tif")
        title = f"{s} Continuous Change Distribution ({d1} to {d2})"
        
        plot_change_continuous(
            raster_path=rpath,
            title=title,
            out_folder=col4,
            season=s,
            d1=d1,
            d2=d2,
            bin_width=1
        )

# -------------------------------------------------------------
# Export summary table
# -------------------------------------------------------------
summary_df = pd.DataFrame(summary_rows)
csv_path = os.path.join(col4, "Directional_Change_Statistics_Table.csv")
summary_df.to_csv(csv_path, index=False)

print("\n📊 Statistical summary table saved to:")
print("   ", csv_path)
print("\n🎯 CELL 4 COMPLETE — PNG figures + statistics table created.")
```
