```python
1 #Wind Direction symbology application


import arcpy
import os

# === Allow ArcPy to overwrite existing outputs ===
arcpy.env.overwriteOutput = True

# === Path to your symbology template (.lyrx) ===
symbology_layer = r"C:\Users\shidebna\Downloads\Thesis_Analysis\Symbology\WindDirection.lyrx"

# === Input → Output folder pairs ===
folders = {
    r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\ten_year_mean":
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\ten_year_mean_symbol",

    r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\seasonal_mean":
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\seasonal_mean_symbol",
    
    r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\New decadal_mean":
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\New decadal_mean_symbol"
}

# === Main processing loop ===
for input_folder, output_folder in folders.items():
    print(f"\n📂 Processing folder: {input_folder}")

    # Ensure output folder exists
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Created output folder: {output_folder}")

    # Set ArcPy workspace
    arcpy.env.workspace = input_folder

    # List all .tif rasters in this folder
    rasters = arcpy.ListRasters("*.tif")

    if not rasters:
        print("⚠ No .tif rasters found in this folder.")
        continue

    # === Process each raster ===
    for raster in rasters:
        in_raster = os.path.join(input_folder, raster)
        temp_layer = os.path.splitext(raster)[0] + "_temp_layer"
        out_layer = os.path.join(
            output_folder, os.path.splitext(raster)[0] + "_symbol.lyrx"
        )

        print(f"→ Applying symbology to: {raster}")

        try:
            # Create a temporary layer from the raster
            arcpy.management.MakeRasterLayer(in_raster, temp_layer)

            # Apply saved symbology (.lyrx template)
            arcpy.management.ApplySymbologyFromLayer(temp_layer, symbology_layer)

            # Save to .lyrx (overwrite enabled)
            arcpy.management.SaveToLayerFile(temp_layer, out_layer, "ABSOLUTE")
            print(f"✔ Exported (or overwritten): {out_layer}")

        except Exception as e:
            print(f"❌ Failed for {raster}: {e}")

        finally:
            # Clean up temporary layer to free memory
            arcpy.management.Delete(temp_layer)

print("\n✅ All .tif rasters processed successfully — symbolized .lyrx files updated.")
```
