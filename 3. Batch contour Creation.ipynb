```python

#contour for change in wind direction

import arcpy, os
from arcpy.sa import *

# === Enable overwriting ===
arcpy.env.overwriteOutput = True

# === Input → Output folder pairs ===
folder_pairs = [
    (
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\multi_interval_change",
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\multi_interval_change_contour"
    ),
    (
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\decadal_change",
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\decadal_change_contour"
    )
]

# === Symbology layer (optional) ===
symbology_layer = r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\decadal_change_symbology\DJF_Change_1985-1994_vs_1995-2004_symbol.lyrx"

# === Contour break values (based on your legend) ===
contour_values = [-20, -15, -10, -5, 0, 5, 10, 15, 20, 180]

# === Process each folder pair ===
for in_folder, out_folder in folder_pairs:
    # Ensure output folder exists
    if not os.path.exists(out_folder):
        os.makedirs(out_folder)
        print(f"📁 Created output folder: {out_folder}")

    # List all .tif files
    rasters = [f for f in os.listdir(in_folder) if f.lower().endswith(".tif")]

    if not rasters:
        print(f"⚠️ No .tif files found in: {in_folder}")
        continue

    print(f"\n=== Processing folder: {in_folder} ===")

    # Loop through rasters
    for raster_name in rasters:
        in_raster = os.path.join(in_folder, raster_name)
        base_name = os.path.splitext(raster_name)[0]
        out_contour = os.path.join(out_folder, f"{base_name}_contour.shp")

        # Delete existing file if it exists (overwrite safeguard)
        if arcpy.Exists(out_contour):
            arcpy.management.Delete(out_contour)
            print(f"♻️ Overwriting existing: {out_contour}")

        try:
            print(f"📈 Generating contour for: {raster_name}")

            # Generate contour lines
            ContourList(
                in_raster=in_raster,
                out_polyline_features=out_contour,
                contour_values=contour_values
            )

            print(f"✅ Contour created: {out_contour}")

            # Apply symbology (if available)
            if arcpy.Exists(symbology_layer):
                arcpy.management.ApplySymbologyFromLayer(
                    in_layer=out_contour,
                    in_symbology_layer=symbology_layer
                )
                print("🎨 Symbology applied.")
            else:
                print("⚠️ Symbology file not found — skipped.")

        except Exception as e:
            print(f"❌ Failed for {raster_name}: {e}")

print("\n✅ All rasters processed successfully — contours saved in their respective folders.")
```

```python
#contour for wind direction mean

import arcpy, os
from arcpy.sa import *

# === Enable Spatial Analyst & Overwrite ===
arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

# === Input → Output Folder Pairs ===
folder_pairs = [
    (
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\ten_year_mean",
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\ten_year_mean_contour"
    ),
    (
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\seasonal_mean",
        r"C:\Users\shidebna\Downloads\Thesis_Analysis\WindDirection_Output\seasonal_mean_contour"
    )
]

# === Contour Break Values (0–360° Compass Intervals) ===
contour_values = [11.24, 33.74, 56.24, 78.74, 101.24, 123.74, 146.24, 168.74,
                  191.24, 213.74, 236.24, 258.74, 281.24, 303.74, 326.24, 348.74, 360]

# === Main Loop ===
for in_folder, out_folder in folder_pairs:
    os.makedirs(out_folder, exist_ok=True)
    rasters = [f for f in os.listdir(in_folder) if f.lower().endswith(".tif")]

    if not rasters:
        print(f"⚠️ No .tif files found in: {in_folder}")
        continue

    print(f"\n=== Processing folder: {in_folder} ===")

    for raster_name in rasters:
        in_raster = os.path.join(in_folder, raster_name)
        base_name = os.path.splitext(raster_name)[0]
        out_contour = os.path.join(out_folder, f"{base_name}_contour.shp")

        # Delete if already exists (overwrite cleanly)
        if arcpy.Exists(out_contour):
            arcpy.management.Delete(out_contour)
            print(f"♻️ Overwriting existing: {out_contour}")

        try:
            print(f"📈 Generating contours for: {raster_name}")

            # Direct contour generation (no intermediate rasters)
            arcpy.sa.ContourList(
                in_raster=in_raster,
                out_polyline_features=out_contour,
                contour_values=contour_values
            )

            print(f"✅ Contour shapefile saved: {out_contour}")

        except Exception as e:
            print(f"❌ Failed for {raster_name}: {e}")

print("\n✅ All rasters processed successfully — only contour shapefiles exported.")
```
